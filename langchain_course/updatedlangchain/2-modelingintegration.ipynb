### Models Integration with OpenAI, Google Gemini and GROQ

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"]= os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
model = init_chat_model("gpt-4.1")
model
```

```python
## invoke the model
response = model.invoke("Hello, How are you?")
response
```

```python
response.content
```

```python
### ChatOpenAI 

from langchain_openai import ChatOpenAI 
ChatOpenAI(model="gpt4.1")
response = model.invoke("Hello How are you?")
response
```

```python
### Google Gemini Models Integration

from langchain.chat_models import init_chat_model
model = init_chat_model("google_genai:gemini-3.5-flash-lite")
response = model.invoke("Why do parrots talk?")
response
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI 

model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
response = model.invoke("Why do Chinese have every possible words in their native langauge?")
response
```

```python
###GROQ Model Integrations

from langchain.chat_models import init_chat_model 

model = init_chat_model("groq:qwen/qwen3.6-27b")
response = model.invoke("Why do fireflies light up?")
response
```

```python
response.content
```

```python
from langchain_groq import ChatGroq

model = ChatGroq(model="qwen/qwen3.6-27b")
response = model.invoke("What is table in chinese called?")
response
```

### Streaming


```python
model.invoke("Write me a 200 words paragraph on AI ")
```

```python
for chunk in model.stream("Write me a 200 words paragraph on PyTorch"):
    print(chunk.text, end="|", flush=True)
```

```python
responses = model.batch([
    "Why do elephants have long nose?",
    "Why do frog hop but not fly?",
    "what is conciousness?"
])

for response in responses:
    print(response)
```

```python
responses = model.batch([
    "Why do elephants have long nose?",
    "Why do frog hop but not fly?",
    "what is conciousness?",
],
    config={
        'max_concurrency': 5   #Limit to 5 parallel calls
    }
)
```
